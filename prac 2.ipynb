```python
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
# Load the pre-trained ResNet50 model + weights trained on ImageNet
model = ResNet50(weights='imagenet')
```

```python
# You can replace 'tomato.jpg' with any image you want to test
img_path = '/content/tomato.jpg'   # Change path as needed
img = image.load_img(img_path, target_size=(224, 224))  # Resize to 224x224 for ResNet50
x = image.img_to_array(img)  # Convert image to array
x = np.expand_dims(x, axis=0)  # Add batch dimension
x = preprocess_input(x)  # Preprocess for ResNet50
```

```python
preds = model.predict(x)
```

```python
print('Predicted:', decode_predictions(preds, top=3)[0])
```
